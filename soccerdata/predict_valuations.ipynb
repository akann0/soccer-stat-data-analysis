```python
# just save the data in a dataframe

with open('../transfermarkt/players.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)
    data = [row for row in reader]

player_df = pd.DataFrame(data)

# I need a dictionary to map player IDs to their names
player_id_to_name = {}
player_name_to_id = {}
for index, row in player_df.iterrows():
    player_id_to_name[row['id']] = row['name']
    player_name_to_id[row['name']] = row['id']
```
