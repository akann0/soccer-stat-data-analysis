```python
def get_stat_tracked(player_line):
    return player_line[PROP_LABEL_STRING].replace(player_line[PARTICIPANT_LABEL_STRING], '').replace("O/U", '').strip().lower()

# choose the value closest to a 50/50 split
# TODO: unders
def decide_to_skip(new_val, old_val):
    new_val_odds = 1 / float(new_val["oddsDecimal"])
    old_val_odds = 1 / float(old_val[new_val['stat_tracked'] + '_' + "Over"])

    new_val_diff = abs(new_val_odds - 0.5)
    old_val_diff = abs(old_val_odds - 0.5)

    return new_val_diff > old_val_diff


    
    
```

```python
final_dk_db = {} #dic so we can combine lines for each player

for index, row in player_lines.iterrows():
    row['stat_tracked'] = get_stat_tracked(row)
    if row[PARTICIPANT_LABEL_STRING] not in final_dk_db:
        final_dk_db[row[PARTICIPANT_LABEL_STRING]] = {
            'name': row[PARTICIPANT_LABEL_STRING]
        }
    

    #TODO: integrate
    value = row['label'].replace("+", '').replace("-", '')
    if not value.isdigit():
        if not row['line'] or not row['line'].replace('.', '', 1).isdigit():
            continue
        over_or_under = "Over" if "Over" in row['label'] else "Under"
        final_dk_db[row[PARTICIPANT_LABEL_STRING]][row['stat_tracked']] = float(row['line'])
        final_dk_db[row[PARTICIPANT_LABEL_STRING]][row['stat_tracked'] + '_' + over_or_under] = row['oddsDecimal']
    else:
        if row['stat_tracked'] in final_dk_db[row[PARTICIPANT_LABEL_STRING]]:
            if (decide_to_skip(row, final_dk_db[row[PARTICIPANT_LABEL_STRING]])):
                continue
        value = float(value)
        over_or_under = "Over" if "+" in row['label'] else "Under"
        value = float(row['label'].replace("+", '').replace("-", ''))
        final_dk_db[row[PARTICIPANT_LABEL_STRING]][row['stat_tracked']] = value - 0.5 if over_or_under == "Over" else value + 0.5
        final_dk_db[row[PARTICIPANT_LABEL_STRING]][row['stat_tracked'] + '_' + over_or_under] = row['oddsDecimal']


final_dk_db = pd.DataFrame(final_dk_db).T
final_dk_db
```

```python
from scipy.stats import poisson
from scipy.optimize import minimize

def implied_odds_at_point(draft_kings_line, lambda_):
        # DraftKings line is an integer threshold (either x.5 or x.0, where x.0 has a push opportunity, so be wary)
        # use the pdf function to get the exact probability of the line
        exact_odds = poisson.pmf(draft_kings_line, lambda_)
        return ((1 - poisson.cdf(draft_kings_line, lambda_)) / (1-exact_odds))


def get_implied_odds(pp_line, draft_kings_line, draft_kings_over_line):
    # Function to solve for lambda
    def solve_lambda(lambda_):
        return abs(implied_odds_at_point(draft_kings_line, lambda_) - (1/draft_kings_over_line))

    # Optimize to find lambda
    # essentially, were setting the two sides of the minus sign to be equal
    result = minimize(solve_lambda, x0=[draft_kings_line], bounds=[(0, None)]) 
    lambda_ = result.x[0]
    print(lambda_, pp_line, draft_kings_line, draft_kings_over_line)
    
    # Calculate the probability of exceeding the pp_line
    prob_over_pp_line = (1 - poisson.cdf(pp_line, lambda_)) / (1 - poisson.pmf(pp_line, lambda_))
    return prob_over_pp_line

results_df["implied_odds"] = results_df.apply(lambda row: get_implied_odds(row["line"], row["dk_line"], row["dk_over_line"]), axis=1)
```
